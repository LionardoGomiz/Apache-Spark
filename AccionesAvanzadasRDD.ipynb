#Almacenamiento en Cache

```python
import findspark

findspark.init()

from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

sc = spark.sparkContext
```

```python
rdd = sc.parallelize([item for item in range(10)])
```

```python
#Importamos la clase/enum 'StorageLevel' desde el paquete 'pyspark.storagelevel'
#Storagelevel define modos de presitencia'
from pyspark.storagelevel import StorageLevel
```

```python
#persist (metodo de RDD): marca el RDD para que, cuando se materialice por primer vez
#(es decir, cuando ejecutemos una ciion, como 'count') se guarde en el nivel indicado
#'StorageLevel.MEMORY_ONLY': Guardar solo en memoria (RAM), mas rapido pero si
#no cabe las partiocnes que se exceden, no se almacenaran y no recalcularan
#rdd.persist(StorageLevel.MEMORY_ONLY)
```

```python
rdd.persist(StorageLevel.DISK_ONLY)
```

```python
rdd.unpersist()
```

```python
rdd.cache()
```

```python
# ----Demo Almacenamiento Cache
rdd.persist
```
